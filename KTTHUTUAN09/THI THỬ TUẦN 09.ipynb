```python
import cv2
import numpy as np
import random


img = cv2.imread('a.jpg')


mean_filtered = cv2.blur(img, (5, 5))
cv2.imwrite('a_mean.jpg', mean_filtered)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 100, 200)
cv2.imwrite('a_edges.jpg', edges)

channels = [0, 1, 2]
random.shuffle(channels)
random_color = img[:, :, channels]
cv2.imwrite('a_random_color.jpg', random_color)


hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv)
cv2.imwrite('a_hue.jpg', h)
cv2.imwrite('a_saturation.jpg', s)
cv2.imwrite('a_value.jpg', v)
```

```python
import cv2
import numpy as np
import random

def inverse(img):
    return 255 - img

def gamma_correction(img):
    gamma = random.uniform(0.5, 2.0)
    inv_gamma = 1.0 / gamma
    table = np.array([(i / 255.0) ** inv_gamma * 255 for i in range(256)]).astype("uint8")
    return cv2.LUT(img, table)

def log_transform(img):
    c = random.uniform(1.0, 5.0)
    log_img = c * np.log1p(img.astype(np.float32))
    return np.clip(log_img, 0, 255).astype(np.uint8)

def hist_equalization(img):
    if len(img.shape) == 2:
        return cv2.equalizeHist(img)
    ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
    ycrcb[:, :, 0] = cv2.equalizeHist(ycrcb[:, :, 0])
    return cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2BGR)

def contrast_stretching(img):
    min_val = random.randint(0, 128)
    max_val = random.randint(128, 255)
    return cv2.normalize(img, None, alpha=min_val, beta=max_val, norm_type=cv2.NORM_MINMAX)

def adaptive_hist_eq(img):
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    lab[:, :, 0] = clahe.apply(lab[:, :, 0])
    return cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)

images = [cv2.imread(f'image{i}.jpg') for i in range(1, 4)]

methods = {
    'I': ('inverse', inverse),
    'G': ('gamma', gamma_correction),
    'L': ('log', log_transform),
    'H': ('hist', hist_equalization),
    'C': ('contrast', contrast_stretching),
    'A': ('adaptive', adaptive_hist_eq)
}


key = input("Nhập phím I/G/L/H/C/A để chọn phương pháp: ").upper()

if key in methods:
    method_name, func = methods[key]
    for i, img in enumerate(images, 1):
        output = func(img)
        filename = f'output_{method_name}_{i}.jpg'
        cv2.imwrite(filename, output)
        print(f'Đã lưu {filename}')
else:
    print("Phím không hợp lệ.")
```

```python
import cv2


img1 = cv2.imread('colorful-ripe-tropical-fruits.jpg')
h1, w1 = img1.shape[:2]
resized1 = cv2.resize(img1, (w1 + 30, h1 + 30))
cv2.imwrite('fruits_resized.jpg', resized1)

img2 = cv2.imread('quang-ninh.jpg')
(h2, w2) = img2.shape[:2]
center = (w2 // 2, h2 // 2)
M = cv2.getRotationMatrix2D(center, -45, 1.0)
rotated = cv2.warpAffine(img2, M, (w2, h2))
flipped = cv2.flip(rotated, 1)  
cv2.imwrite('quangninh_rotated_flipped.jpg', flipped)


img3 = cv2.imread('pagoda.jpg')
enlarged = cv2.resize(img3, None, fx=5, fy=5, interpolation=cv2.INTER_LINEAR)
blurred = cv2.GaussianBlur(enlarged, (7, 7), 0)
cv2.imwrite('pagoda_blurred.jpg', blurred)

```
